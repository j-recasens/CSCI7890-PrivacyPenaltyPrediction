## Considerations: Defining a Unified Taxonomy

**The challenge:** EU and US labels operate at different levels of the legal stack.
- EU labels = specific GDPR articles (legal provisions, often 3-5 per case)
- US labels = federal statutes (one or two per case, coarser grain)

These are not directly comparable as-is. A unified taxonomy requires mapping both onto a common set of violation *dimensions*.

**Dr. Ramasamy's 6-dimension framework (from PPKG paper) as a candidate:**
| Dimension | EU analog | US analog |
|---|---|---|
| Disclosure | Articles 13, 14 | Section 5 (deceptive) |
| Lawful Basis / Consent | Articles 6, 7 | COPPA |
| Third-Party Transparency | Article 13(1)(e), 28 | Privacy Shield, GLBA |
| Security | Article 32 | GLBA Safeguards, HBNR, FCRA |
| Rights | Articles 17, 21 | FCRA (dispute rights) |
| Risk / General | Article 83 | Section 5 Only |

**Open questions for meeting:**
- Her framework was designed for privacy policy text (what companies promise), not enforcement decisions (what regulators found). Does the mapping still hold?
- Do we adopt her 6 dimensions as our unified label set and recode both corpora?
- Or do we keep parallel native labels and treat cross-regime transfer as the research question rather than forcing a common taxonomy?
