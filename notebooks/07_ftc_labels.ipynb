```python
import urllib.request
import json
import os
import pandas as pd
```

```python
RAW = "/Users/nic/Documents/MM2/data/raw/ftc_rafal_cases.json"
os.makedirs(os.path.dirname(RAW), exist_ok=True)

url = "https://www.rafalsportfolio.me/data/ftc-cases.json"
req = urllib.request.Request(url, headers={"User-Agent": "PrivacyEnforcementResearch/0.1(jr38088@georgiasouthern.edu)"})
with urllib.request.urlopen(req, timeout=60) as r, open(RAW, "wb") as f:
    f.write(r.read())

with open(RAW) as f:
    data = json.load(f)
print(data["total_cases"], "cases | generated_at:", data["generated_at"])
    
```

```python
df = pd.DataFrame(data["cases"])

list_cols = [c for c in df.columns if df[c].map(lambda v: isinstance(v, list)).any()]
for c in list_cols:
    df[c] = df[c].map(lambda v: "; ".join(map(str, v)) if isinstance(v, list) else v)
print("flattened list columns:", list_cols)

df["provision_counts_by_topic"] = df["provision_counts_by_topic"].map(lambda v: json.dumps(v) if isinstance(v, dict) else v)

df.to_csv("/Users/nic/Documents/MM2/data/ftc_rafal_cases.csv", index=False)
print(df.shape)
df.head()
```
